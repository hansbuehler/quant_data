```python
df = pd.read_csv("results/best_params_test.csv")
```

```python
df
```

```python
# Average puts & calls together: mean of every metric per (date, scale)
group_cols = ["model", "side"]
metric_cols = df.select_dtypes("number").columns.difference(group_cols)

combined = df.groupby(group_cols, as_index=False)[list(metric_cols)].mean()
```

```python
combined[
    ["model", "ho_nse_mean", "ho_nse_med", "ho_ese_mean", "ho_out_pct_mean", "arb_frac", "side"]
].sort_values("ho_nse_mean", ascending=True)
```
