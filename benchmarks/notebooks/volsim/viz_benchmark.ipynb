```python
%load_ext autoreload
%autoreload 1

%aimport openqai.core
%aimport openqai.core.biz_calendar
%aimport openqai.data.widb
%aimport openqai.data.widb.client
%aimport openqai.data.widb.day
%aimport openqai.data.widb.arbfree
%aimport openqai.data.widb.options_filter
%aimport openqai.core.backbone.backbone
%aimport openqai.benchmarks.sanos.sanos_tools
%aimport openqai.benchmarks.sanos.arbfree
%aimport openqai.benchmarks.sanos.sanos
%aimport openqai.benchmarks.volsim.benchmark_grid

from openqai.benchmarks.volsim.benchmark_grid import GridMarketDefinition, DateRange, PoolConfig, Context, date, GridConfig

print("*** Loading data")
grid_config     = GridConfig(grid_num_strikes=100)
grid_market_def = GridMarketDefinition( ticker="SPX", date_range=DateRange(start=date(2025, 1, 1), end=date(2025, 2, 1)), grid_config=grid_config )
market          = grid_market_def.load( pool_config = PoolConfig(num_workers=4, threading=False ), verbose=Context.all, override_cache_mode="off" )
print(f"nms /done - found {len(market['dates'])} dates.")
    
```

```python
def plot():
    from prettytable import PrettyTable
    from cdxcore.util import fmt_big_byte_number

    tbl = PrettyTable(align="l", field_names=['Field', 'Shape', 'Dtype', 'Memory'])
    size = 0
    for k, v in market.items():
        if k == "ticker":
            tbl.add_row([k, v, '', ''])
            continue
        if k in ['grid_market_def']:
            continue
        size += v.nbytes
        tbl.add_row([k, v.shape, v.dtype, fmt_big_byte_number(v.nbytes)])
    
    tbl.add_row(["Total", '', '', fmt_big_byte_number(size)])
    display(tbl)

plot()
```
