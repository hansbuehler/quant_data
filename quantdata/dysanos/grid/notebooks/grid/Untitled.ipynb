```python
import torch
import numpy as np
from dataclasses import dataclass

x = torch.tensor( np.random.normal(size=(10,2)) )
y = torch.tensor( np.random.normal(size=(10,2)) )

class C:

    def __init__(self, s, x):
        self.s = s
        self.x = x

    def __call__(self, y):
        return self.s*self.x*y

def f(s,x):
    return C(s=s,x=x)

def g(s,x,y):
    c = f(s,x)
    return c(y), c

g = torch.compile(g)
g(0.1,x,y)
```
