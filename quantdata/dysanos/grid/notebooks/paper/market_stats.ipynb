```python
import numpy as np
from prettytable import PrettyTable
from cdxcore.dynaplot import figure

def stats(ex):
    spots  = ex.universe['spots']
    dS_sim = np.diff(spots[:,1:]/spots[:,:-1] - 1., axis=1 ).reshape(-1)
    dt_sim = ex.universe['dt']
    assert isinstance(dt_sim, float), f"Expected dt_sim to be a float but got {type(dt_sim  )}"

    dS_mkt = ex.market['spot_ret'][1:]
    time_y = ex.market['time_y']
    dt_mkt = np.diff(time_y, axis=0).reshape(-1)
    assert dt_mkt.shape == dS_mkt.shape, f"Expected dt_mkt shape {dt_mkt.shape} to match dS_mkt shape {dS_mkt.shape}"

    def comp(dS,dt):
        tbl = PrettyTable()
        tbl.field_names = ["mean", "std", "median", "q10", "q90"]
        q10, med, q90 = np.quantile(dS/dt, (0.1, 0.5, 0.9))
        tbl.add_row([ f"{np.mean(dS/dt):.2%}", f"{np.std(dS/np.sqrt(dt)):.2%}", f"{med:.2%}", f"{q10:.2%}", f"{q90:.2%}" ])
        return tbl

    print( "Simulated dS stats:\n", comp(dS_sim, dt_sim), "\nMarket dS stats:\n", comp(dS_mkt, dt_mkt) )

    with figure(figsize=(8, 4)) as fig:
        ax1 = fig.add_subplot()
        ax1.hist(dS_sim/dt_sim, bins=50, alpha=0.7, weights=np.ones_like(dS_sim/dt_sim) / len(dS_sim/dt_sim), edgecolor='lightblue', color="blue", label="Simulated")
        ax1.hist(
            dS_mkt/dt_mkt,
            bins=50,
            weights=np.ones_like(dS_mkt/dt_mkt) / len(dS_mkt/dt_mkt),
            alpha=0.7,
            edgecolor='lightgreen',
            color="green",
            label="Market"
        )
        ax1.set_title('Market Returns')
        ax1.set_xlabel('Return')
        ax1.set_ylabel('Frequency')
        ax1.legend(loc="upper right")

    

stats(ex_orig)
stats(ex_all_alpha)

```
