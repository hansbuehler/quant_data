```python
from quantdata.dysanos.grid.torch_drift_change import DualRiskNeutralDeepHedgingConfig, FeedForwardConfig, train, TrainConfig, VizConfig, BaseEnvironment, Context
import torch as torch

mdl_config = DualRiskNeutralDeepHedgingConfig( 
    label="25k-ent1-revision",
    group="dysanos", 
    user_version="0.1", 
    utility="entropy", 
    lmnda=1.0, 
    delta_hedged=False, 
    dual_weight=0.1,
    tert_weight=0.1,
    spot_only=False,
    cost = 0.0001, # 1 bps; dual cost is 10 times higher
    boot_strap=100,
    boot_strap_dual=100,
    boot_strap_tert=100,
    use_fixed=True,
    in_config=FeedForwardConfig(
        num_depth = 2,
        num_width = 20,
        activation = "LeakyReLU",
        use_bias = True
    ),
    pf_config=FeedForwardConfig(
        num_depth = 2,
        num_width = 50, # we have ~140 instruments per time step
        activation = "LeakyReLU",
        use_bias = True
    ),
    _debug_no_xstate = False,
    _debug_ignore_prior = False,
    _debug_do_not_normalize_dQdP = False
)

import gc
gc.collect()

model_RN_output = 0
gpu_id = 0
device = torch.device(f"cuda:{gpu_id}" if torch.cuda.is_available() else "cpu")

def run():
    trn_config = TrainConfig( num_epochs=10_000, 
                              pcnt_validation=0.1,
                              lr=1E-2, final_lr=1E-6, weight_decay=1e-4, batch_size=20_000, compile=False, epoch_interval=10 )
    viz_config = VizConfig()
    model_RN   = train(
                    ex.universe,
                    ex.sampler, 
                    ref_dQdP=ex.dQdP_dS,
                    mdl_config=mdl_config, 
                    trn_config=trn_config, 
                    viz_config=viz_config, restart=False, just_load="no",
                   env=BaseEnvironment(debug=False), verbose=Context.all )
    gc.collect()
    return model_RN

model_RN_output = run()
```
