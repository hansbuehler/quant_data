```python
from quantdata.dysanos.grid.torch_drift_change import DualRiskNeutralDeepHedgingConfig, FeedForwardConfig, _notebook_name, train, TrainConfig, VizConfig, BaseEnvironment, Context
import torch as torch

mdl_config = DualRiskNeutralDeepHedgingConfig( 
    label=_notebook_name(),
    group="dysanos", 
    user_version="0.2", 
    utility="entropy", 
    lmnda=1.0, 
    delta_hedged=False, 
    dual_weight=0.1,
    tert_weight=0.1,
    cost = 0.000, # 1 bps
    boot_strap=1,
    boot_strap_dual=1,
    boot_strap_tert=1,
    use_fixed=True,
    in_config=FeedForwardConfig(
        num_depth = 2,
        num_width = 40,
        activation = "LeakyReLU",
        use_bias = True
    ),
    pf_config=FeedForwardConfig(
        num_depth = 2,
        num_width = 40,
        activation = "LeakyReLU",
        use_bias = True
    ),
)

import gc
model_RN_output = 0
gc.collect()
print(torch.cuda.memory_allocated() / 1024**3, "GiB allocated")
print(torch.cuda.memory_reserved() / 1024**3, "GiB reserved")

def run():
    trn_config = TrainConfig( num_epochs=200_000, 
                              pcnt_validation=0.,
                              lr=1E-2, final_lr=1E-6, weight_decay=0., batch_size=None, compile=True, epoch_interval=100 )
    viz_config = VizConfig()
    model_RN   = train(
                    ex.universe,
                    ex.sampler, 
                    ref_dQdP=ex.dQdP_dS,
                    mdl_config=mdl_config, 
                    trn_config=trn_config, 
                    viz_config=viz_config, restart=False, just_load="no",
                   env=BaseEnvironment(debug=False), verbose=Context.all )
    gc.collect()
    return model_RN

model_RN_output = run()
```
