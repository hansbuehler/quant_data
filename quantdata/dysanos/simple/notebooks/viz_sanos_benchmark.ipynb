```python


%load_ext autoreload
%autoreload 1

%aimport quantdata
%aimport quantdata.biz_calendar
%aimport quantdata.utils.arbfree
%aimport quantdata.widb
%aimport quantdata.widb.client
%aimport quantdata.widb.day
%aimport quantdata.widb.arbfree
%aimport quantdata.widb.options_filter
%aimport quantdata.widb.timeseries
%aimport quantdata.dysanos.simple.sanos_benchmark

from datetime import date
from quantdata.widb.timeseries import build_time_series, TimeSeriesDefinition, ExpiryFilter, RawDayFilter, OptionsFilter, DateRange, PoolConfig, table_of_time_series, MkArbFree
from quantdata.biz_calendar import next_business_day
from cdxcore.util import fmt_date, fmt_dict
from cdxcore.verbose import Context
from IPython.display import display

ticker = "SPX"
time_series_def = TimeSeriesDefinition(
    raw_day_filter = RawDayFilter(),
    expiry_filter = ExpiryFilter(),
    mk_arb_free = MkArbFree(),
    options_filter = OptionsFilter()
)
date_range = DateRange(start=date(2025, 5, 1), end=date(2025, 7, 1))
pool_config = PoolConfig(num_workers=2, threading=True)

samples = build_time_series(ticker, date_range, include_fields=None, exclude_fields=None, time_series_def=time_series_def, pool_config=pool_config, verbose= Context.all, override_cache_mode="on" )

verbose = Context.all
verbose.write(str(table_of_time_series(samples)))
print("/done")
```

```python
from quantdata.dysanos.simple.sanos_benchmark import SanosBenchmarkConfig, sanos_benchmark


def benchmark(samples):
    benchmark_config = SanosBenchmarkConfig()
    pool_config = PoolConfig(num_workers=0, threading=True)
    r = sanos_benchmark( ticker=ticker, date_range=date_range, 
                         time_series=samples,time_series_def=time_series_def, 
                         benchmark_config=benchmark_config, pool_config=pool_config, 
                         verbose=Context.all, override_cache_mode="off")

bm = benchmark(samples)
```
