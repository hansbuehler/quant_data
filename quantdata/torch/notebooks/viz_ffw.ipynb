## Testing FFW

```python
%load_ext autoreload
%autoreload 1

%aimport quantdata.torch.ffw

from cdxcore.verbose import Context
verbose = Context("all")
with verbose.write_t("Importing torch...", end='') as tme:   
    import torch
    verbose.write(f"done; this took {tme}.", head=False)

from quantdata.torch.ffw import FeedForward, FeedForwardConfig, BaseEnvironment

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

N = 3
env = BaseEnvironment(device=device, debug=True)
ffw_config = FeedForwardConfig(
    num_depth = 1,
    num_width = 32,
    activation = "ReLU",)
x = torch.randn(2,3).to(env.device)
ffw = FeedForward( env=env, num_input=N, num_output=1, config=ffw_config, use_final_bias=True, use_final_activation=False )
y = ffw(x)
print(y.detach().cpu(), y.device)

ffw_config = FeedForwardConfig(
    num_depth = 1,
    num_width = 32,
    activation = "Softplus",)
ffw = FeedForward( env=env, num_input=N, num_output=1, config=ffw_config, use_final_bias=True, use_final_activation=False )
x = torch.ones((10, N), dtype=torch.float32 ).to(env.device)
x = torch.randn(2,3).to(env.device)
y = ffw(x)
print(y.detach().cpu(), y.device)


```
