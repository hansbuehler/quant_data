```python
%load_ext autoreload
%autoreload 1

%aimport quantdata
%aimport quantdata.biz_calendar
%aimport quantdata.massive.client
%aimport quantdata.massive.options.contracts

from datetime import date
import polars as pl
from quantdata.massive.client import MassiveClient
from quantdata.pool import PoolConfig
from quantdata.massive.options.contracts import ContractsFilter, filter_contracts, filter_contracts_between, Context, REJECTED, DateRange
from IPython.display import display, Markdown
```

```python

pool_config = PoolConfig(num_workers=8, threading=False)
contracts_filter = ContractsFilter(max_dte=25, max_mns=3.0, min_opt_vol_k=1)

df = filter_contracts(
    "SPY",
    date(2024, 1, 2),
    contracts_filter=contracts_filter,
    exclude_half_days=True,
    override_cache_mode="off",
    pool_config=pool_config,
    verbose=Context.all,
)
display(Markdown("## All contracts"))
display(df)
display(df.filter(pl.col("ticker").str.ends_with("500")))

display(Markdown("## Active contracts"))
display(df.filter(pl.col("reject_code") == REJECTED.USED))
```
