## Test WIDB client

```python
import os
os.environ["QUANTDATA_RAW_DATA"] = "!/quantdata/raw" # better set these to environment variables
os.environ["QUANTDATA_TRAINING_DATA"] = "!/quantdata/training" # better set these to environment variables
os.environ["QUANTDATA_PROCESSING_DATA"] = "!/quantdata/processing" # better set these to environment variables
os.environ["QUANTDATA_CACHE_MODE"] = "off" # turns off global cache mode

%load_ext autoreload
%autoreload 1

%aimport quantdata
%aimport quantdata.biz_calendar
%aimport quantdata.widb.client

from datetime import date
from quantdata.widb.client import WIDBClient
from quantdata.biz_calendar import next_business_day
from cdxcore.util import Mapping, fmt_date, fmt_dict
from cdxcore.verbose import Context
from IPython.display import display

def test():
    ticker = "SPX"
    d      = next_business_day( date(2024, 3, 3) )
    client = WIDBClient()
    secid  = client.lookup_secid(ticker)

    verbose = Context.all

    with verbose.write_t(f"Loading zero curve for {d}...", end='') as tme:
        zero_curve = client.zero_curve(d)
        verbose.write(f"done; this took {tme}. Found {len(zero_curve)} rows.", head=False)

    with verbose.write_t(f"Loading spot price for {ticker} on {d}...", end='') as tme:
        spot_price = client.spot_price(secid, d)
        verbose.write(f"done: spot price for {ticker} on {fmt_date(d)} is {spot_price}; this took {tme}.", head=False)

    with verbose.write_t(f"Loading forward curve for {d}...", end='') as tme:
        forwards = client.forwards(secid, d)
        verbose.write(f"done; this took {tme}. Found {len(forwards)} rows.", head=False)

    with verbose.write_t(f"Loading options for {d}...", end='') as tme:
        options = client.options(secid, d)
        verbose.write(f"done; this took {tme}. Found {len(options)} rows.", head=False)

    display(zero_curve)
    display(forwards)
    display(options)    

test()
```
