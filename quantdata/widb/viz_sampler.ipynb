```python

import os
os.environ["QUANTDATA_CACHE_MODE"] = "off" # turns off global cache mode; useful for testing but do not use in general

%load_ext autoreload
%autoreload 1

%aimport quantdata
%aimport quantdata.biz_calendar
%aimport quantdata.widb
%aimport quantdata.widb.client
%aimport quantdata.widb.day
%aimport quantdata.widb.sampler

from datetime import date
from quantdata.widb.client import WIDBClient
from quantdata.widb.sampler import sample_days, ExpiryFilter, RawDayFilter, SampleDaysConfig, DateRange, PoolConfig, table_of_sample_days
from quantdata.biz_calendar import next_business_day
from cdxcore.util import Mapping, fmt_date, fmt_dict
from cdxcore.verbose import Context
from IPython.display import display

def test():
    ticker = "SPX"
    date_range = DateRange(start=date(2025, 1, 1), end=date(2025, 3, 1))
    raw_day_filter = RawDayFilter()
    expiry_filter = ExpiryFilter()
    sample_days_config = SampleDaysConfig()
    pool_config = PoolConfig(num_workers=4, threading=True) # no parallel processing for testing
    verbose = Context.all

    samples = sample_days(ticker, date_range, include_fields=None, exclude_fields=None, raw_day_filter=raw_day_filter, expiry_filter=expiry_filter, sample_days_config=sample_days_config, pool_config=pool_config, verbose=verbose )
    return samples

samples = test()
verbose = Context.all
verbose.write(str(table_of_sample_days(samples)))
print("/done")
```
